```python
df.replace("?", np.nan, inplace=True)
```

```python
missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0]

print(missing_values.sort_values(ascending=False))
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage = missing_percentage[missing_percentage > 0]

print(missing_percentage.sort_values(ascending=False))
```

```python
plt.figure(figsize=(12, 6))

missing_percentage.sort_values(ascending=False).plot(kind='bar')

plt.title("Missing Values Percentage by Attribute")
plt.xlabel("Attributes")
plt.ylabel("Missing Values (%)")
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

```python
print("Duplicate records:", df.duplicated().sum())
```

```python
print("Total encounters:", len(df))

print("Unique patients:", df["patient_nbr"].nunique())
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df, x="age", order=sorted(df["age"].dropna().unique()))

plt.title("Distribution of Patients by Age")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df, x="age", hue="readmitted")

plt.title("Readmission Distribution Across Age Groups")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(data=df, x="time_in_hospital", bins=14)

plt.title("Distribution of Time Spent in Hospital")
plt.xlabel("Days in Hospital")
plt.ylabel("Number of Encounters")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(data=df, x="number_inpatient")

plt.title("Distribution of Previous Inpatient Visits")
plt.xlabel("Number of Previous Inpatient Visits")
plt.ylabel("Number of Encounters")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=["int64", "float64"])

plt.figure(figsize=(12, 8))

sns.heatmap(numeric_df.corr(), cmap="coolwarm", annot=False)

plt.title("Correlation Heatmap of Numerical Features")

plt.show()
```
